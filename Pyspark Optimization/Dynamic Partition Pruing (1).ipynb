```python
# Turn off Adaptive Query Execution (AQE)
spark.conf.set("spark.sql.adaptive.enabled", "false")

# Turn off Dynamic Partition Pruning (DPP)
spark.conf.set("spark.sql.optimizer.dynamicPartitionPruning.enabled", "false")

# Turn off automatic broadcast joins
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", -1)
```

```python
df = spark.read.format("csv")\
    .option("header", True)\
    .option("inferSchema", True)\
    .load("/FileStore/rawdata/BigMart_Sales.csv")

df.limit(8)
```

**this 8 data we can use cause in this item id is unique n can be used as PK**

**Preparing partoition data**


```python
df.write.format("parquet")\
    .mode("append")\
    .partitionBy("Item_identifier")\
    .option("path", "/FileStore/rawdata/dpp_partitionednew")\
    .save()
```

3 file will be created in partition

**Non  Partition data**

```python
df.write.format("parquet")\
    .mode("append")\
    .option("path", "/FileStore/rawdata/dpp_Nonpartitioned")\
    .save()
```

###Data frame


```python
df1 = spark.read.format("parquet")\
    .load("/FileStore/rawdata/dpp_partitioned")

df2 = spark.read.format("parquet")\
    .load("/FileStore/rawdata/dpp_partitioned")
```

```python
# Import all PySpark SQL functions
from pyspark.sql.functions import *

# Join df1 with filtered records from df2
# Filter df2 where Item_Identifier is "FDA15"
# Join using Item_Identifier as the common column
# Use an inner join to keep only matching records
df_join = df1.join(
    df2.filter(col("Item_Identifier") == "FDA15"),
    df1["Item_Identifier"] == df2["Item_Identifier"],
    "inner"
)

# Display the joined DataFrame
df_join.display()
```
