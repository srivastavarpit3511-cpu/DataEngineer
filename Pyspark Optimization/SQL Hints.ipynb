```python
from pyspark.sql.functions import *
spark.conf.set("spark.sql.adaptive.enabled", "false")
```

```python
# Big DataFrame
df_transactions = spark.createDataFrame([
    (1, "US", 100),
    (2, "IN", 200),
    (3, "UK", 150),
    (4, "US", 80)
], ["id", "country_code", "amount"])


# Small DataFrame
df_countries = spark.createDataFrame([
    ("US", "United States"),
    ("IN", "India"),
    ("UK", "United Kingdom")
], ["country_code", "country_name"])
```

```python
df_transactions.createOrReplaceTempView("transactions")
df_countries.createOrReplaceTempView("countries")
 
```

```python
df_sql=spark.sql('''select * from transactions t  join
 countries c on t.country_code=c.country_code''')
```

```python
df_sql.display()
```

###sql hint

```python
df_sql_opt=spark.sql('''select * /*broadcast(c) */ from transactions t  join
 countries c on t.country_code=c.country_code''')
```
