```python
from pyspark.sql.functions import *
```

```python

# Creating a DataFrame
data = [("A", 100), ("A", 200), ("A", 300), ("B", 400), ("C", 500)]

df = spark.createDataFrame(data, ["user_id", "purchase"])

df.display()
```

```python
df = df.withColumn("salt_column", floor(rand() * 3))

df.display()
```

**Creating Concat Column on original groupBy col and salt_column to create a new groupBy col**

```python
df = df.withColumn(
    "user_id_salt",
    concat(col("user_id"), lit("_"), col("salt_column"))
)

df.display()
```

####### Applying Group By on the new column

```python
df = df.groupBy("user_id_salt").agg(sum("purchase"))

df.display()
```
